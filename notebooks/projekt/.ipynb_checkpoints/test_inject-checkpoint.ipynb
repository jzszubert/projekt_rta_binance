```python
from kafka import KafkaProducer
import json, time

p = KafkaProducer(bootstrap_servers="broker:9092",
                  value_serializer=lambda v: json.dumps(v).encode())
p.send("klines", {
    "symbol": "BTCUSDT", "start_time": int(time.time() * 1000),
    "open_price": "66800", "high_price": "66900", "low_price": "66700",
    "close_price": "66830", "volume": "999",
    "taker_buy_volume": "500", "is_closed": True
})
p.flush()
```
