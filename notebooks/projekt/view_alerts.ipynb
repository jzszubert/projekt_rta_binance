```python
from kafka import KafkaConsumer
import json
c = KafkaConsumer("alerts", bootstrap_servers="broker:9092",
                  auto_offset_reset="earliest",
                  value_deserializer=lambda v: json.loads(v.decode()),
                  consumer_timeout_ms=10000)
for m in c:
    print(json.dumps(m.value, indent=2, ensure_ascii=False))
c.close()
```
